```python
import pandas as pd


url = "/content/car.data"

columns = [
    "buying",
    "maint",
    "doors",
    "persons",
    "lug_boot",
    "safety",
    "class"
]

data = pd.read_csv(url, names=columns)


print("First 5 rows of the dataset:")
print(data.head())




data["target"] = data["class"].apply(
    lambda x: "yes" if x == "vgood" else "no"
)



def find_s(data):

    features = [
        "buying",
        "maint",
        "doors",
        "persons",
        "lug_boot",
        "safety"
    ]


    hypothesis = ["0"] * len(features)

    print("\nFind-S Learning Process:")
    print("-" * 50)

    for index, row in data.iterrows():


        if row["target"] == "yes":

            print("\nPositive Example:", index + 1)

            for i, feature in enumerate(features):


                if hypothesis[i] == "0":
                    hypothesis[i] = row[feature]


                elif hypothesis[i] != row[feature]:
                    hypothesis[i] = "?"

            print("Hypothesis:", hypothesis)

    return hypothesis



final_hypothesis = find_s(data)


print("\nFinal Hypothesis:", final_hypothesis)
print("\n" + "=" * 50)
print("FINAL HYPOTHESIS")
print("=" * 50)

for feature, value in zip(columns[:-1], final_hypothesis):
    print(feature, "=", value)
```
