```python
import pandas as pd

df = pd.read_csv('/content/tennis.csv')

x = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

hypothesis = None
for i in range(len(x)):
  if y[i] == 'Yes':
    if hypothesis is None:
      hypothesis = x[i].copy()
    else:
      for j in range(len(hypothesis)):
        if hypothesis[j] != x[i][j]:
          hypothesis[j] = '?'
  print("step", i+1,  "hypothesis" , hypothesis)
print("\n Final Hypothesis: " , hypothesis)
```
