```python
import pandas as pd

data = {
    'Age': [20, 21, 22, 23, 24, 25, 40, 41, 42, 43, 44, 45],
    'Spending_Score': [80, 85, 78, 82, 88, 90, 30, 25, 28, 35, 32, 27]
}

df = pd.DataFrame(data)

df
```

```python
import matplotlib.pyplot as plt

plt.scatter(df['Age'], df['Spending_Score'])

plt.xlabel('Age')
plt.ylabel('Spending Score')
plt.title('Customer Data')

plt.show()
```

```python

X = df[['Age', 'Spending_Score']]

X
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

kmeans.fit(X)
```

```python

df['Cluster'] = kmeans.labels_

df

```

```python

plt.scatter(
    df['Age'],
    df['Spending_Score'],
    c=df['Cluster']
)

plt.xlabel('Age')
plt.ylabel('Spending Score')
plt.title('K-Means Clustering')

plt.show()
```

```python

print("Cluster Centers:")
print(kmeans.cluster_centers_)


```

```python

plt.scatter(
    df['Age'],
    df['Spending_Score'],
    c=df['Cluster']
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker='X',
    s=200
)

plt.xlabel('Age')
plt.ylabel('Spending Score')
plt.title('K-Means Clustering with Centroids')

plt.show()

```

```python


for cluster in sorted(df['Cluster'].unique()):
    print(f"\nCluster {cluster}:")
    print(df[df['Cluster'] == cluster])

```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

for k in range(1, 7):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

plt.plot(range(1, 7), inertia, marker='o')

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.title('Elbow Method')

plt.show()



```

```python

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)

df['Cluster'] = kmeans.fit_predict(X)

df

plt.scatter(
    df['Age'],
    df['Spending_Score'],
    c=df['Cluster']
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker='X',
    s=200
)

plt.xlabel('Age')
plt.ylabel('Spending Score')
plt.title('Final K-Means Clusters')

plt.show()
```
